 Key Findings

1. Attack Distribution is Nearly Balanced

* The dataset contains **40,000 cyber attack records**.
* Attack types are almost evenly distributed:

  * DDoS: **13,428 attacks**
  * Malware: **13,307 attacks**
  * Intrusion: **13,265 attacks**
* No single attack type dominates, indicating organizations must prepare for multiple threat categories.



2. High-Severity Threats Represent Significant Risk

* High Severity: **13,382 attacks**
* Medium Severity: **13,435 attacks**
* Low Severity: **13,183 attacks**
* High and medium severity attacks account for nearly two-thirds of total incidents, requiring continuous monitoring and rapid response mechanisms.



3. Network Traffic Uses Multiple Attack Protocols

* ICMP: **13,429 attacks**
* UDP: **13,299 attacks**
* TCP: **13,272 attacks**
* Attack traffic is evenly distributed across protocols, suggesting attackers leverage diverse communication channels to bypass security controls.



4. All Network Segments Are Equally Exposed

* Segment A: **13,273 attacks**
* Segment B: **13,319 attacks**
* Segment C: **13,408 attacks**
* Cyber threats affect all network segments at similar levels, indicating organization-wide security measures are necessary rather than focusing on a single segment.



5. Attack Sources Are Highly Distributed

* Top source IP addresses generated only one recorded attack each.
* No single IP address appears repeatedly as a major threat source.
* This suggests distributed attack behavior, making traditional IP-blocking strategies less effective.



6. Attack Targets Are Widely Distributed

* Destination IP analysis shows attacks are spread across multiple target systems.
* No single destination IP dominates attack volume.
* Security monitoring should cover the entire network infrastructure rather than a few critical assets.



7. Geographic Attack Sources Are Diverse

* Top locations account for only 13–16 attacks each.
* No clear geographic hotspot exists.
* Threats originate from multiple regions, emphasizing the need for global threat intelligence and monitoring.


8. Security Environment Shows Constant Threat Activity

* Balanced attack types, severity levels, protocols, and network segments indicate a continuously active threat environment.
* Organizations should adopt layered cybersecurity defenses including:

  * IDS/IPS monitoring
  * Firewall protection
  * Anomaly detection systems
  * Incident response procedures

Business Recommendations

1. Implement real-time threat monitoring across all network segments.
2. Prioritize detection of high and medium severity attacks.
3. Strengthen protocol-level monitoring for TCP, UDP, and ICMP traffic.
4. Deploy anomaly-based detection instead of relying solely on IP blocking.
5. Conduct regular security assessments to identify emerging vulnerabilities.
6. Maintain centralized logging and incident response mechanisms.
