Purpose
Prepare data for analysis.
This notebook fixes messy data

```python
import pandas as pd
import os
print(os.getcwd())
```

```python
df = pd.read_csv("/Users/artitakle/Documents/Cyber Security Attack Analysis/Data/Raw/cybersecurity_attacks.csv")
```

Removing Duplicates

```python
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print(df)
```

Checking Null Values

```python
print(df.isnull().sum())
```

Filling Missing Values

```python
df['Malware Indicators'] = df['Malware Indicators'].fillna('No Malware Detected')
df['Alerts/Warnings'] = df['Alerts/Warnings'].fillna('No Alerts/Warnings')
df['Proxy Information'] = df['Proxy Information'].fillna('No Proxy Information')
df['Firewall Logs'] = df['Firewall Logs'].fillna('No Firewall Logs')
df['IDS/IPS Alerts'] = df['IDS/IPS Alerts'].fillna('No IDS/IPS Alerts')
```

Verfying Null values

```python
print(df.isnull().sum())
```

```python
print(df.dtypes)
```

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
```

```python
df.shape[0]
```

```python
output_dir = "data/processed"
os.makedirs(output_dir, exist_ok=True)
df.to_csv(os.path.join(output_dir, "cleaned_attacks.csv"), index=False)
print("Cleaned  dataset saved successfully.")
```
